```python
import pandas as pd

df1 = pd.DataFrame(
    {"acc": [13, 21, 33, 48], "name": ["Ram", "Shyam", "Hari", "Krishna"], "age" : [21, 65, 34, 88]} 
)

df2 = pd.DataFrame(
    {"acc": [13, 21, 33, 48], "principle": [500, 658, 4032, 500], "interest" : [6.3, 4.25, 9.8, 5.6]} 
)

```

```python
df1, df2
```

# Concatenation

```python
pd.concat([df1, df2], axis = 1)
```

```python
pd.concat([df1, df2], axis = 0, ignore_index = True)
```

# Merging

```python
import pandas as pd

df1 = pd.DataFrame(
    {"key": ["AB", "CB", "AC", "DE"], "name": ["Ram", "Shyam", "Hari", "Krishna"], "age" : [21, 65, 34, 88]} 
)

df2 = pd.DataFrame(
    {"key": ["AB", "CC", "AC", "DF"], "principle": [500, 658, 4032, 500], "interest" : [6.3, 4.25, 9.8, 5.6]} 
)
```

```python
df1, df2
```

```python
pd.merge(df1, df2, how = "inner", on = "key")
```

```python
pd.merge(df1, df2, how = "right", on = "key")
```

```python
import pandas as pd

df1 = pd.DataFrame(
    {"key": ["AB", "CB", "AC", "DE"], "name": ["Ram", "Shyam", "Hari", "Krishna"], "age" : [21, 65, 34, 88]} 
)

df2 = pd.DataFrame(
    {"key": ["AB", "CC", "AC", "DF"], "principle": [500, 658, 4032, 500], "age" : [6.3, 4.25, 9.8, 5.6]} 
)
```

```python
df1, df2
```

```python
pd.merge(df1, df2, how = "inner", on = "key")
```

```python
pd.merge(df1, df2, how = "inner", on = "key", suffixes = ("_df1", "_df2"))
```

```python
# Alternative syntax

df1.merge(df2, on = "key", how = "inner")
```
