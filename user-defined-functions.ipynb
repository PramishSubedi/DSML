```python
import math

def standard_deviation(numbers):
    mean = sum(numbers) / len(numbers)

    total = 0

    for number in numbers:
        total = total + (number - mean) ** 2

    result = math.sqrt(total / len(numbers))

    return result


numbers = [10, 20, 30, 40, 50]

print(round(standard_deviation(numbers), 2))
```

```python
def calculate_std(*numbers):

    N = len(numbers)
    mean = sum(numbers) / N

    numerator = sum((number - mean) ** 2 for number in numbers)

    std = (numerator / N) ** 0.5 

    return round(std, 2)
```

```python
calculate_std(1,2,3,4,5)
```

```python
def power_up(base, power):
    return base ** power
```

```python
power_up(2,2)
```

```python
power_up_lambda = lambda base, power : base ** power
```

```python
power_up_lambda(2,2)
```

```python
area = lambda l, b, h: 2 * (l*b + b*h + l*h)

print(area(10, 5, 3))
```

```python
def area_cuboid(l, b, h) -> float:

    area = 2 * (l * b + b * h + l * h)
    return area
```

```python
area_cuboid(2,2,2)
```
