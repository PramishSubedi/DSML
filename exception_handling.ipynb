```python
lst = ["a","b","c","d"]
 
def get_elements(index):
    return lst[index]
    
try:
    user_index = int(input("Index of list"))
    print(get_elements(user_index))

except IndexError:
    print(get_elements(-1))
 
# If index value  > len(lst) --> Error
# In that condition , return last element of the list
```

```python
def division():

    try:
        num1 = int(input("Enter the first number: "))
        num2 = int(input("Enter the second number: "))

        result = num1 / num2

        print(f"The division of two numbers is: {result}")

    except ZeroDivisionError as e:
        print(f"Cannot divide by zero: {e}")


division()
```

```python
def integer():
    try:
        value = input("Enter an integer: ")
        value = int(value)

        print(value)

    except ValueError:
        print("Please enter a valid integer.")


integer()
```
