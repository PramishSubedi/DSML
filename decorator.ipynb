```python
def decorator(func): #function as argument/parameter


    def wrapper():

        print("Before Function Call")
        print(func().upper()) #actual function call
        print("After Function Call")

    return wrapper #return function as value
```

```python
@decorator
def greet():
    return "Hello, World"
```

```python
greet()
```

```python
import time


def get_time(func):
    def wrapper():
        start = time.time()

        func()

        end = time.time()
        print("Time taken:", end - start, "seconds")

    return wrapper


@get_time
def hello():
    time.sleep(2)
    print("Hello")


hello()
```
