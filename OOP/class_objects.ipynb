# CLASS AND OBJECTS

```python
class Calculator:

    def __init__(self, num1, num2):
        print("Making Onjects.....")
        self.num1 = num1
        self.num2 = num2


    def add(self, x, y):
        return x+y

    
    def multiply(self, num3: int) -> int | float:
        return self.num1 * self.num2 * num3
```

```python
calc = Calculator(2,3)
```

```python
calc.num2
```

```python
calc.multiply(2)
```

# TYPES OF ATTRIBUTES

- INSTANCE ATTRIBUTE|S

- CLASS ATTRIBUTES

```python
class Dog:
    count = 0 #<- Class Attibute

    def __init__(self, name: str, age: int = 1):
        #Initializer or Constructor
        self.name = name #<-instance attribute
        self.age = age #<-instance attribute

        Dog.count += 1
```

```python
dog1 = Dog("Luffy", 3)
```

```python
dog1.name, dog1.age
```

```python
dog1.count
```

```python
dog2 = Dog("Jojo", 7)
```

```python
dog2.name, dog2.age
```

```python
dog2.count
```
