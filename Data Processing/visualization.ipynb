```python
ax = titanic_df.Survived.value_counts().plot( 
    kind = "bar",
    title = "Count of Survived Passengers",
    figsize = (4,4),
    legend = True,
    xlabel = "Survived (0 = No, 1 = Yes)",
    ylabel = "No. of Passengers",
    grid = False,
    yticks = range(0, 551, 50),
    colormap = "Accent",
    #color = ["red", "green"]
)

ax.bar_label(ax.containers[0], padding = 0)
```

```python
titanic_df.value_counts(subset = ["Sex", "Survived"]).plot( 
    kind = "bar",
    title = "Count of Survived Passengers",
    figsize = (4,4),
    legend = True,
    xlabel = "Survived (0 = No, 1 = Yes)",
    ylabel = "No. of Passengers",
    grid = False,
    yticks = range(0, 551, 50),
    colormap = "Accent",
    #color = ["red", "green"]
)
```

```python
crosstab = pd.crosstab(
    titanic_df.Survived,
    titanic_df.Pclass,
)
crosstab
```

```python
crosstab = pd.crosstab(
    titanic_df.Survived,
    titanic_df.Sex,
).plot( 
    kind = "bar",
    title = "Count of Survived Passengers",
    figsize = (4,4),
    legend = True,
    xlabel = "Survived (0 = No, 1 = Yes)",
    ylabel = "No. of Passengers",
    grid = False,
    yticks = range(0, 551, 50),
    colormap = "tab20",
    #color = ["red", "green"]
)
```
