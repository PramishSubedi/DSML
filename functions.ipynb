all(), any() FUNCTIONS

```python
list_bool = [False, True, True, False, 0]
```

```python
any(list_bool) #returns true if any of the iterable are true
```

```python
all(list_bool) #returns true if all of the iterable are true
```

```python
#conetxt manager
with open("my.txt", mode = "r+") as file:
    print(file.read())
    file.write("\n Hello Pramish")

print("Closed")
```

enumerate() FUNCTION

```python
ls = ["a", "b", "c"] 
res = enumerate(ls)
```

```python
list(res)
```

```python
for i, name in enumerate(ls, start=1):
    print(f"{name} - Roll No. {i}")
```
