map(), filter() and reduce()

```python
def get_cube(num):
    return num ** 3
```

```python
ls = [1,2,3,4,5]
```

```python
map_obj = map(get_cube, ls)
```

```python
map_obj
```

```python
list(map_obj)
```

```python
list(map_obj)
```

```python
filter_obj = filter(lambda num: num % 2 == 0, ls)
```

```python
filter_obj
```

```python
list(filter_obj)
```

RECURSIVE FUNCTION

```python
def fib(n):
    #base case
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        #recursive case
        return fib(n-1) + fib(n-2)
```

```python
%%time
fib(8)
```

```python
def factorial(n):
    if n == 1:
        return 1
    else:
        return n * factorial(n - 1)
    

factorial(5)
```
