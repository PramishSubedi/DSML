```python
fruits = ['pramish', 'subedi', '22', 'age', 'baglung']

i = 0

while i < len(fruits):
    fruit = fruits[i]

    if fruit == "22":
        i += 1      # Move to the next item before continue
        continue

    print(fruit)
    i += 1
```

```python
n = int(input("how many Fibonacci numbers do you want?:  "))

a = 0
b = 1
count = 0

while count < n:
    print(a)
    c = a + b
    a = b
    b = c
    count += 1

```

Dictionary Comprehension

```python
squared = { i : i ** 2  for i in range(1,21) if i % 2 == 0 }
squared
```

```python
sent = "Although she was extremely tired after working a long shift at the local hospital, Maria still managed to stop by the grocery store, pick up fresh vegetables for dinner, and help her younger brother with his math homework."

res = { word : len(word) for word in sent.split() if len(word)>4 }
res
```

Set Comprehension

```python
squared = { num ** 2  for num in range(1,11) }
squared
```

```python
chars = ["c", "a", "b"]
pairs = { char1 + char2 for char1 in chars for char2 in chars }
pairs
```

###Problem: Validate and Clean User Input Stream

raw_data = ['150', 'invalid', '80', '204', '999', 'STOP', '300', '450']
 
Viral-Load should be in between [100, 999]
 
Expected Output: [150, 204, 300, 450, 999]
 

```python
raw_data = ['150', 'invalid', '80', '204', '999', 'STOP', '300', '450']

clean_data = [ int(item) for item in raw_data if item.isdigit() and 100 <= int(item) <= 999 ] 

print(clean_data)
```
