```python
marks = int(input("Enter your marks"))

if marks < 0:
    print("Invalid marks! Marks cannot be negative.")

elif marks > 100:
    print("Invalid marks! Marks cannot be more than 100.")

elif marks >= 90:
    print("You have got Grade A+")

elif marks >= 80:
    print("You have got Grade A")

elif marks >= 70:
    print("You have got Grade B+")

elif marks >= 60:
    print("You have got Grade B")

elif marks >= 50:
    print("You have got Grade C+")

elif marks >= 40:
    print("You have got Grade C")

else:
    print("You have Failed.")
```

LOOPING STATEMENT

```python
fruits = ["apple", "mango", "strawberry"]

for fruit in fruits:
    print(f"{fruit:} -Rs.100")
```

```python
for i in range(5):
    print("Pramish")
```

```python
for i in range (1, 10):
    if i == 5:
        break
    print(i)
```

```python
for i in range (1, 10):
    if i == 5:
        continue
    print(i)
```

```python
count = 1

while count <=5:
    print(count)
    count +=1
```

```python
for i in range(1, 6):
    print("*" * i)
```

```python
adj = ["red", "big", "tasty"]
fruits = ["apple", "mango", "strawberry"]

for fruit in fruits:
    for ad in adj:
        print(ad, fruit)
```

```python
number = 8

for i in range(1, 11):
    print(f"{number} x {i} = {number * i}")
```
