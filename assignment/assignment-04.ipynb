# Math Tutor Using Random, Arithmetic Operators, and OOP

Scenario: Create a MathTutor class that generates random math questions using random and math operators.

- Track correct and incorrect answers.
- Provide a score at the end.
- Handle invalid input using exception handling.

Concepts: Random, Math, Exception Handling, Classes

```python
import random


class MathTutor:
    def __init__(self):
        self.correct = 0
        self.incorrect = 0

    def generate_question(self):
        num1 = random.randint(1, 10)
        num2 = random.randint(1, 10)

        operator = random.choice(["+", "-", "*", "/"])

        # question = f"{num1} {operator} {num2}"
        # answer = eval(question)
        # return question, answer

        if operator == "+":
            answer = num1 + num2

        elif operator == "-":
            answer = num1 - num2

        elif operator == "*":
            answer = num1 * num2

        elif operator == "/":
            answer = num1 / num2

        return num1, operator, num2, answer

    def ask_question(self):

        num1, operator, num2, correct_answer = self.generate_question()

        while True:
            try:
                user_answer = float(input(f"What is {num1} {operator} {num2}?"))

                if user_answer == correct_answer:
                    print("Correct")
                    self.correct += 1

                else:
                    print(f"Sorry! The correct answer is {correct_answer}")
                    self.incorrect += 1

                break

            except ValueError:
                print("Invalid Input! Please enter a number.")

    def show_score(self):
        total = self.correct + self.incorrect
        print("\n-------Final Score-------")
        print("Correct:", self.correct)
        print("Incorrect:", self.incorrect)
        print("Total Questions:", total)

        if total > 0:
            score = (self.correct / total) * 100
            print(f"Score: {score:.2f}%")
```

```python
tutor = MathTutor()
```

```python
for i in range(5):
    tutor.ask_question()
```

```python
tutor.show_score()
```

# Exception Handling Scenario: Online Age-Restricted Service

Scenario: You’re building a sign-up system for an online movie rental platform. Some movies are age-restricted (18+). You need to ensure proper validation and error handling during user registration.

Task:

1. Create a custom exception class called UnderageError that inherits from Exception.

2. Write a function register_user() that:

- Takes a user’s name and age as input.
- Raises UnderageError if the user is under 18.
- Otherwise, prints a welcome message.

3. Wrap the function call in a try block and handle the exception.

4. Use else to confirm successful registration and finally to always print “Thank you for using MovieTime!” regardless of outcome.

Also try to validate if the age input is numeric. Raise a ValueError if not, and handle it separately.

```python
import time


class UnderageError(Exception):
    pass

def register_user(name, age):
    if age >= 18:
        print(f"Welcome, {name}")

    else:
        raise UnderageError("Must be an adult to register. Age must be 18 or above.")

    
try:
    name = input("Enter your name: ")
    age = int(input("Enter your age: "))

    print("Verifying age! Wait a second \n")
    time.sleep(3)

    register_user(name, age)

except ValueError as e:
    print("Please enter a valid numeric age:", e)
        
except UnderageError as e:
    print("Registration failed:", e)

else:
    print("You have been successfully registered!")

finally:
    print("Thank you for using MovieTime!")
```

# Problem Statement

We want to build an online shopping cart system that allows users to add products to their cart, calculate the total cost, apply discounts, and generate an invoice. The system should include the following functionalities:

- Adding products to the cart
- Removing products from the cart
- Calculating the total cost
- Applying discounts based on user type
- Generating an invoice

1. Create the Product class

We create a basic Product class with attributes for the product name and price.

```python
class Product:

  def __init__(self, name, price):
    self.name = name
    self.price = price


class User:

  def __init__(self, users_name, is_admin=False):
    self.users_name = users_name
    self.is_premium = False
    self.is_admin = is_admin  
    self.cart = ShoppingCart(self)

  def make_premium(self, user):
    if self.is_admin:
      user.is_premium = True
      print(f"{user.users_name} is now a premium member.")
    else:
      print("Only an admin can make a user premium.")

  def new_admin(self, user):
    if self.is_admin:
      user.is_admin = True
      print(f"{user.users_name} is now an admin.")
    else:
      print("Only an admin can create another admin.")


class ShoppingCart:

  def __init__(self, user):
    self.user = user
    self.products = []

  def add_product(self, product):
    self.products.append(product)

  def remove_product(self, product):
    if product in self.products:
      self.products.remove(product)
    else:
      print("Product is not in the cart.")

  def calculate_total_cost(self):
    cost = sum(product.price for product in self.products)
    if self.user.is_premium:
      discount = cost * 10 / 100
      total_cost = cost - discount
    else:
      total_cost = cost
    return total_cost


class Invoice:

  def __init__(self, cart):
    self.cart = cart

  def generate_invoice(self):
    cost = sum(product.price for product in self.cart.products)
    total_cost = self.cart.calculate_total_cost()
    discount = cost - total_cost

    print("\n========== INVOICE ==========")
    print(f"Customer: {self.cart.user.users_name}")
    print(f"Premium Member: {self.cart.user.is_premium}")

    print("\nProducts:")
    for product in self.cart.products:
      print(f"{product.name}: Rs. {product.price}")

    print(f"\nSubtotal: Rs. {cost}")
    print(f"Discount: Rs. {discount}")
    print(f"Total: Rs. {total_cost}")
    print("=============================")


```

```python
# testing

product1 = Product("Laptop", 100000)
product2 = Product("Mouse", 2000)
product3 = Product("Keyboard", 3000)

user1 = User("Pramish")
user2 = User("Prajwol")
admin = User("Admin", is_admin = True)


# each user has their own independent cart

user1.cart.add_product(product1)
user1.cart.add_product(product2)

user2.cart.add_product(product2)
user2.cart.add_product(product3)


# removing product from Pramish's cart

user1.cart.remove_product(product2)


# admin makes Prajwol a premium user

admin.make_premium(user2)


# admin creates another admin

admin.new_admin(user1)


# calculating total cost

print(f"Pramish's total cost: Rs. {user1.cart.calculate_total_cost()}")
print(f"Prajwol's total cost: Rs. {user2.cart.calculate_total_cost()}")


# generating invoices

invoice1 = Invoice(user1.cart)
invoice1.generate_invoice()

invoice2 = Invoice(user2.cart)
invoice2.generate_invoice()
```
