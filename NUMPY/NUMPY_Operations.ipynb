solve equation: 4x + 2y = 8 , 5x + 3y =9

```python
A = np.array([[4,2], [5,3]])
B = np.array([8,9])
```

```python
A_inv = np.linalg.inv(A)
```

```python
X = np.dot(A_inv, B)
X
```
